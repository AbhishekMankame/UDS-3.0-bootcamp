A string in Python is a sequence of characters enclosed within single quotes(' '), double quotes (" "), or triple quotes (''' ''' or """ """). Strings are used to store and represent text, such as names, sentences, numbers written as text, and symbols.

#### strip() in Python
The `strip()' method in Python is used to remove unwanted whitespaces or specified characters from the beginning (left side) and end (right end) of a string. By default, it removes spaces, tabs (\t), and new line characters (\n) from both ends of a string. <br>
It does not remove spaces or characters present in the midding of a string. Since strings are immutable in Python, the strip() method returns a new string without chaning the original string.

#### Types of strip methods:

Python provides thress commonly used methods:
1. strip(): Removes whitespace or specified characters from both the beginning and end of a string.
2. lstrip(): Removes whitespace or specified characters from the left side (beginning) of a string.
3. rstrip(): Removes whitespace or specifed characters from the right side (end) of a string.
